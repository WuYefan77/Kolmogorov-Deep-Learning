# Task 1 — SDE simulation and convergence
Validate Euler–Maruyama for geometric Brownian motion against the exact transition. We examine coupled strong error and weak moment error.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
import matplotlib.pyplot as plt
from src.sde import convergence_tables
```

```python
strong, weak = convergence_tables(n_paths=60_000)
print(strong)
print(weak)
```

```python
plt.figure(figsize=(6,4))
plt.loglog(strong.dt, strong.strong_rmse, 'o-', label='EM strong RMSE')
plt.loglog(strong.dt, strong.strong_rmse.iloc[-1]*(strong.dt/strong.dt.iloc[-1])**0.5, '--', label='slope 1/2')
plt.xlabel('dt'); plt.ylabel('error'); plt.legend(); plt.grid(True, which='both', alpha=.25);
```

```python
w2 = weak[weak.moment_power==2]
plt.figure(figsize=(6,4))
plt.loglog(w2.dt, w2.weak_abs_error, 'o-', label='weak error for S^2')
plt.loglog(w2.dt, w2.weak_abs_error.iloc[-1]*(w2.dt/w2.dt.iloc[-1]), '--', label='slope 1')
plt.xlabel('dt'); plt.ylabel('error'); plt.legend(); plt.grid(True, which='both', alpha=.25);
```
