# Task 2 — Feynman–Kac Monte Carlo
Estimate a European call from exact GBM simulations and verify the expected Monte Carlo RMSE rate N^{-1/2}.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
import matplotlib.pyplot as plt
from src.monte_carlo import black_scholes_call, mc_convergence_experiment, mc_call_price
```

```python
truth = float(black_scholes_call(100, 100, 0.05, 0.2, 1.0))
mc = mc_convergence_experiment(reps=60)
print('Black–Scholes truth:', truth)
print(mc)
```

```python
plt.figure(figsize=(6,4))
plt.loglog(mc.n_paths, mc.rmse, 'o-', label='empirical MC RMSE')
plt.loglog(mc.n_paths, mc.rmse.iloc[0]*(mc.n_paths/mc.n_paths.iloc[0])**(-0.5), '--', label='slope -1/2')
plt.xlabel('number of paths'); plt.ylabel('RMSE'); plt.legend(); plt.grid(True, which='both', alpha=.25);
```

```python
estimate, se = mc_call_price(100, 100, 0.05, 0.2, 1.0, 100_000, seed=42, antithetic=True)
print(f'antithetic estimate = {estimate:.6f}, SE = {se:.6f}, 95% CI = [{estimate-1.96*se:.6f}, {estimate+1.96*se:.6f}]')
```
